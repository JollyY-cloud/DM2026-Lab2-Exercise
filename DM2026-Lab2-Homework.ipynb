### 1.1 Preprocessing Steps

**Syntax:** `###` creates a tertiary heading (H3).

I started by combining three Kaggle datasets: `final_posts.json`,`data_identification.csv`, and `emotion.csv`, using `post_id` as the
matching key. The dataset contains 4,800 tweets, including 3,360 training samples and 1,440 test samples.

I first checked missing values, class distribution, duplicate texts, and conflicting labels. There were no missing texts or training labels.
However, the six emotion classes were imbalanced: joy was the largest class (875 samples), while fear was the smallest (288 samples).

I also found 50 duplicate-text rows in the training set, 16 unique texts with conflicting emotion labels,
and 59 unique texts shared between the training and test sets. 
I kept the original data unchanged and used these findings to guide validation and later experiments.

To reduce data leakage during validation, I used `StratifiedGroupKFold` with five splits and a random seed of 42.
Identical texts were grouped together so they could not appear in both training and validation sets.

The resulting split contained 2,688 training samples and 672 validation samples, with zero overlapping texts. 
This provided a consistent evaluation setup for comparing different models.

### 1.2 Feature Engineering Steps

I first used TF-IDF to represent the tweets for traditional machine learning models. 
The word-level features included unigrams and bigrams, with `min_df=2` and sublinear term frequency scaling. This generated 6,025 features.

I then added character-level TF-IDF using 3–5 character n-grams. Combining word and character features increased the total feature count to 48,714. 
This helped capture spelling variations and short text patterns commonly found in tweets.

For the Transformer models, I used the pretrained tokenizer from `SamLowe/roberta-base-go_emotions`, with a maximum sequence length of 128 tokens. 
The six emotion labels were encoded as integers from 0 to 5.

To avoid information leakage, the TF-IDF vocabulary was fitted only on the training subset. 
The validation data was transformed using the same vocabulary.

Adding character-level features improved the Linear SVM validation Macro F1 from 0.4543 to 0.4918, 
suggesting that character n-grams provided useful additional information.

### 1.3 Explanation of Your Model

I started with a Linear SVM classifier using TF-IDF features as my baseline. 
The word-level model achieved a validation Macro F1 of 0.4543, which improved to 0.4918 after combining word and character n-grams. 
However, the performance was still limited, especially for emotions such as disgust.

To improve classification performance, I experimented with pretrained Transformer models, including RoBERTa, DistilRoBERTa, and DeBERTa-v3-large with LoRA fine-tuning. 

The emotion-aware RoBERTa model achieved a validation Macro F1 of 0.6336, outperforming the traditional SVM baseline.

I then combined five models using weighted probability averaging. 
The original three-model group contained two RoBERTa models with different random seeds and one DistilRoBERTa model. 
The newer group included emotion-aware RoBERTa and DeBERTa-v3-large LoRA. Giving the two groups weights of 40% and 60% produced a Macro F1 of 0.6527.

Since the dataset had imbalanced emotion classes, I also tested class-balanced Focal Loss with γ = 1.5. Although the Focal Loss model performed slightly worse on its own (Macro F1 = 0.6270), it provided some complementary predictions. Adding it to the five-model ensemble with a 15% weight improved Macro F1 to 0.6559 and accuracy to 0.6741.

However, the improvement was small, and the paired bootstrap confidence interval included zero, so I cannot conclude that the six-model ensemble was significantly better.

For the final Kaggle submission, I used the six-model ensemble to predict the 1,440 test samples. I also applied an exact-text matching correction using training tweets with consistent labels, which changed six predictions. The final submission was saved as `submission_six_focal15_exactmatch.csv`.

### STEP 1 | Load and inspect Kaggle datasets

```python
from pathlib import Path
import pandas as pd
from IPython.display import display

# Set the original dataset directory
data_path = Path("data/kaggle")

# Required Kaggle files
required_files = [
    "final_posts.json",
    "data_identification.csv",
    "emotion.csv",
    "sampleSubmission.csv"
]

# Check whether all files are available
for filename in required_files:
    filepath = data_path / filename
    print(f"{filename}: {'Found' if filepath.exists() else 'Missing'}")

# Load datasets only if all files exist
if all((data_path / name).exists() for name in required_files):
    posts = pd.read_json(data_path / "final_posts.json")
    identification = pd.read_csv(data_path / "data_identification.csv")
    emotion = pd.read_csv(data_path / "emotion.csv")
    sample = pd.read_csv(data_path / "sampleSubmission.csv")

    print("\nDataset shapes:")
    print("Posts:", posts.shape)
    print("Identification:", identification.shape)
    print("Emotion:", emotion.shape)
    print("Sample submission:", sample.shape)

    print("\nTrain/Test distribution:")
    display(identification["set"].value_counts())
```

### STEP 2 | Inspect dataset columns and examples

```python
datasets = {
    "Posts": posts,
    "Identification": identification,
    "Emotion": emotion,
    "Sample Submission": sample
}

for name, df in datasets.items():
    print(f"\n{'=' * 45}")
    print(f"{name}")
    print(f"{'=' * 45}")

    print("Columns:", df.columns.tolist())
    print("Shape:", df.shape)
    print("Missing values:")
    print(df.isna().sum())

    print("\nFirst 3 rows:")
    display(df.head(3))
```

### STEP 3 | Data Merging

```python
# STEP 3 | Merge Kaggle datasets
# Combine tweet text, train/test information, and emotion labels

# 1. Verify required columns
assert {"post_id", "text"}.issubset(posts.columns)
assert {"post_id", "set"}.issubset(identification.columns)
assert {"post_id", "label"}.issubset(emotion.columns)

# 2. Merge text with train/test identification
df_merged = posts.merge(
    identification,
    on="post_id",
    how="inner",
    validate="one_to_one"
)

# 3. Separate training and test data
train_df = df_merged[df_merged["set"] == "train"].copy()
test_df = df_merged[df_merged["set"] == "test"].copy()

# 4. Merge training labels
train_df = train_df.merge(
    emotion,
    on="post_id",
    how="left",
    validate="one_to_one"
)

# 5. Validate data integrity
assert len(df_merged) == 4800
assert len(train_df) == 3360
assert len(test_df) == 1440
assert train_df["label"].notna().all()

print("Merged dataset:", df_merged.shape)
print("Training dataset:", train_df.shape)
print("Testing dataset:", test_df.shape)

print("\nMissing training labels:")
print(train_df["label"].isna().sum())

print("\nTraining examples:")
display(train_df.head(3))

print("\nTesting examples:")
display(test_df.head(3))

print("\nData merging completed successfully.")
```

### STEP 4 | Data Quality & Emotion Distribution

```python
# STEP 4 | Data quality and emotion distribution
# Inspect the original text and labels without modifying them

# 1. Emotion distribution
print("=== Emotion Distribution ===")

label_counts = train_df["label"].value_counts()

emotion_summary = pd.DataFrame({
    "Count": label_counts,
    "Percentage (%)": label_counts / len(train_df) * 100
}).round(2)

display(emotion_summary)

# 2. Missing values
print("\n=== Missing Values ===")
print("Train missing text:", train_df["text"].isna().sum())
print("Test missing text:", test_df["text"].isna().sum())
print("Train missing labels:", train_df["label"].isna().sum())

# 3. Duplicate texts
print("\n=== Duplicate Texts ===")

print(
    "Train duplicate text rows:",
    train_df.duplicated(subset=["text"]).sum()
)

print(
    "Test duplicate text rows:",
    test_df.duplicated(subset=["text"]).sum()
)

# 4. Conflicting labels for identical text
conflicting_texts = (
    train_df.groupby("text")["label"]
    .nunique()
    .gt(1)
    .sum()
)

print("\n=== Conflicting Labels ===")
print("Texts with multiple labels:", conflicting_texts)

# 5. Exact text overlap between Train and Test
train_texts = set(train_df["text"].dropna())
test_texts = set(test_df["text"].dropna())

overlapping_texts = train_texts.intersection(test_texts)

print("\n=== Train/Test Text Overlap ===")
print("Unique overlapping texts:", len(overlapping_texts))
```

### STEP 5 | Group-aware Training / Validation Split

```python
# STEP 5 | Group-aware Training / Validation Split

from sklearn.model_selection import StratifiedGroupKFold
import pandas as pd

# 1. Prepare working datasets
# Keep the original data unchanged
train_clean = train_df[["post_id", "text", "label"]].copy()
test_clean = test_df[["post_id", "text"]].copy()

labels = [
    "anger", "disgust", "fear",
    "joy", "sadness", "surprise"
]

# 2. Create a group-aware split
sgkf = StratifiedGroupKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

train_idx, val_idx = next(
    sgkf.split(
        train_clean["text"],
        train_clean["label"],
        groups=train_clean["text"]
    )
)

model_train = train_clean.iloc[train_idx].copy()
model_val = train_clean.iloc[val_idx].copy()

# 3. Check for duplicate text leakage
shared_texts = (
    set(model_train["text"])
    & set(model_val["text"])
)

assert len(model_train) + len(model_val) == len(train_clean)
assert len(shared_texts) == 0
assert set(model_train["post_id"]).isdisjoint(
    set(model_val["post_id"])
)

# 4. Display results
print("=== Dataset Summary ===")
print("Full Training:", len(train_clean))
print("Model Training:", len(model_train))
print("Validation:", len(model_val))
print("Shared Texts:", len(shared_texts))

print("\n=== Emotion Distribution ===")

distribution = pd.DataFrame({
    "Training": model_train["label"].value_counts(),
    "Validation": model_val["label"].value_counts()
}).reindex(labels, fill_value=0)

display(distribution)

print("\nTrain/Validation split completed.")
 
```

## 2. Feature Engineering Steps

### STEP 6 | Word-level TF-IDF Feature Engineering

```python
# STEP 6 | Word-level TF-IDF Feature Engineering

from sklearn.feature_extraction.text import TfidfVectorizer

# 1. Create TF-IDF vectorizer
tfidf = TfidfVectorizer(
    ngram_range=(1, 2),
    min_df=2,
    sublinear_tf=True
)

# 2. Fit on training data only
X_train = tfidf.fit_transform(model_train["text"])

# 3. Transform validation data
X_val = tfidf.transform(model_val["text"])

# 4. Prepare labels
y_train = model_train["label"]
y_val = model_val["label"]

# 5. Check feature dimensions
print("Training features:", X_train.shape)
print("Validation features:", X_val.shape)
print("Vocabulary size:", len(tfidf.vocabulary_))
print("Training labels:", y_train.shape)
print("Validation labels:", y_val.shape)
```

### STEP 7 | Character-level TF-IDF Feature Engineering

```python
# STEP 7 | Character-level TF-IDF Feature Engineering

from scipy.sparse import hstack

# 1. Create character-level TF-IDF
char_tfidf = TfidfVectorizer(
    analyzer="char",
    ngram_range=(3, 5),
    min_df=2,
    sublinear_tf=True
)

# 2. Fit on training data only
X_train_char = char_tfidf.fit_transform(model_train["text"])
X_val_char = char_tfidf.transform(model_val["text"])

# 3. Combine word-level and character-level features
X_train_combined = hstack(
    [X_train, X_train_char]
).tocsr()

X_val_combined = hstack(
    [X_val, X_val_char]
).tocsr()

# 4. Verify feature dimensions
print("Word features:", X_train.shape[1])
print("Character features:", X_train_char.shape[1])

print("\nCombined Training:", X_train_combined.shape)
print("Combined Validation:", X_val_combined.shape)

assert X_train_combined.shape[1] == X_val_combined.shape[1]

print("\nFeature engineering completed successfully.")
```

### STEP 8 | Baseline Model: Word TF-IDF + Linear SVM

```python
# STEP 8 | Baseline Model: Word TF-IDF + Linear SVM

from sklearn.svm import LinearSVC
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    classification_report
)

# 1. Initialize the baseline classifier
svm_word = LinearSVC(
    C=1.0,
    random_state=42,
    max_iter=5000
)

# 2. Train using word-level TF-IDF features
svm_word.fit(X_train, y_train)

# 3. Predict validation labels
pred_word = svm_word.predict(X_val)

# 4. Evaluate model performance
word_accuracy = accuracy_score(y_val, pred_word)
word_macro_f1 = f1_score(
    y_val,
    pred_word,
    average="macro"
)

print("=== Word TF-IDF + Linear SVM ===")
print(f"Accuracy: {word_accuracy:.4f}")
print(f"Macro F1: {word_macro_f1:.4f}")

print("\n=== Classification Report ===")
print(
    classification_report(
        y_val,
        pred_word,
        labels=labels,
        digits=4,
        zero_division=0
    )
)
```

### STEP 9 | Word + Character TF-IDF with Linear SVM

```python
# STEP 9 | Word + Character TF-IDF with Linear SVM

from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score, f1_score, classification_report

# 1. Train with combined TF-IDF features
svm_combined = LinearSVC(
    C=1.0,
    random_state=42,
    max_iter=5000
)

svm_combined.fit(X_train_combined, y_train)

# 2. Predict validation labels
pred_combined = svm_combined.predict(X_val_combined)

# 3. Evaluate the model
combined_accuracy = accuracy_score(y_val, pred_combined)
combined_macro_f1 = f1_score(
    y_val,
    pred_combined,
    average="macro"
)

print("=== Word + Character TF-IDF + Linear SVM ===")
print(f"Accuracy: {combined_accuracy:.4f}")
print(f"Macro F1: {combined_macro_f1:.4f}")

print("\n=== Comparison with Word-only TF-IDF ===")
print(f"Word-only Macro F1: {word_macro_f1:.4f}")
print(f"Combined Macro F1: {combined_macro_f1:.4f}")
print(f"Macro F1 Difference: {combined_macro_f1 - word_macro_f1:+.4f}")

print("\n=== Classification Report ===")
print(
    classification_report(
        y_val,
        pred_combined,
        labels=labels,
        digits=4,
        zero_division=0
    )
)
```

### STEP 12 | Transformer Tokenization & Dataset Preparation

```python
# STEP 14 | Emotion-aware RoBERTa Fine-tuning
# Original competition experiment performed in Google Colab

import numpy as np
import torch

from sklearn.metrics import f1_score, accuracy_score
from transformers import (
    AutoModelForSequenceClassification,
    TrainingArguments,
    Trainer,
    DataCollatorWithPadding,
    set_seed
)

# 1. Verify the training environment
assert torch.cuda.is_available(), "GPU is required for this training experiment."

set_seed(42)

# 2. Load the pretrained classification model
emotion_model = AutoModelForSequenceClassification.from_pretrained(
    checkpoint,
    num_labels=6,
    label2id=label2id,
    id2label=id2label,
    problem_type="single_label_classification",
    ignore_mismatched_sizes=True,
    dtype=torch.float32
)

# 3. Define evaluation metrics
def compute_emotion_metrics(eval_pred):
    logits, true_labels = eval_pred
    predicted = np.argmax(logits, axis=1)

    return {
        "macro_f1": f1_score(
            true_labels,
            predicted,
            average="macro"
        ),
        "accuracy": accuracy_score(
            true_labels,
            predicted
        )
    }

# 4. Configure the original training experiment
emotion_args = TrainingArguments(
    output_dir="/content/goemotions_training",
    num_train_epochs=3,
    learning_rate=2e-5,
    per_device_train_batch_size=8,
    per_device_eval_batch_size=16,
    gradient_accumulation_steps=2,
    weight_decay=0.01,
    eval_strategy="epoch",
    save_strategy="epoch",
    save_total_limit=1,
    load_best_model_at_end=True,
    metric_for_best_model="macro_f1",
    greater_is_better=True,
    logging_strategy="epoch",
    fp16=False,
    bf16=False,
    seed=42,
    data_seed=42,
    report_to="none"
)

# 5. Initialize the Trainer
emotion_trainer = Trainer(
    model=emotion_model,
    args=emotion_args,
    train_dataset=emotion_train,
    eval_dataset=emotion_val,
    processing_class=emotion_tokenizer,
    data_collator=DataCollatorWithPadding(
        tokenizer=emotion_tokenizer
    ),
    compute_metrics=compute_emotion_metrics
)

# 6. Train the model and evaluate validation performance
emotion_trainer.train()

result = emotion_trainer.predict(emotion_val)

y_true = np.asarray(emotion_val["labels"])
y_pred = np.argmax(result.predictions, axis=1)

macro_f1 = f1_score(y_true, y_pred, average="macro")
accuracy = accuracy_score(y_true, y_pred)

print("=== RoBERTa Validation Results ===")
print(f"Macro F1: {macro_f1:.4f}")
print(f"Accuracy: {accuracy:.4f}")

print("\nBest checkpoint:")
print(emotion_trainer.state.best_model_checkpoint)
```

### STEP 17 | Class-balanced Focal Loss Implementation

```python
# STEP 17 | Class-balanced Focal Loss Implementation
# Based on the completed Google Colab experiment

import numpy as np
import pandas as pd
import torch
import torch.nn.functional as F
from transformers import Trainer

# 1. Calculate training class frequencies
class_counts = np.bincount(
    model_train["label"].map(label2id).to_numpy(),
    minlength=6
)

# 2. Calculate class-balanced weights
class_weights = np.sqrt(
    len(model_train) / (6 * class_counts)
)

class_weights = class_weights / class_weights.mean()

# 3. Define the original Focal Loss Trainer
class FocalTrainer(Trainer):

    def compute_loss(
        self,
        model,
        inputs,
        return_outputs=False,
        num_items_in_batch=None
    ):
        targets = inputs["labels"]

        model_inputs = {
            k: v for k, v in inputs.items()
            if k != "labels"
        }

        outputs = model(**model_inputs)
        logits = outputs.logits.float()

        ce = F.cross_entropy(
            logits,
            targets,
            reduction="none"
        )

        pt = torch.exp(-ce)

        weights = torch.as_tensor(
            class_weights,
            dtype=logits.dtype,
            device=logits.device
        )

        focal_loss = (
            weights[targets]
            * (1 - pt).pow(1.5)
            * ce
        ).mean()

        return (
            (focal_loss, outputs)
            if return_outputs
            else focal_loss
        )

# 4. Verify class weights
assert len(class_weights) == 6
assert np.isfinite(class_weights).all()
assert (class_weights > 0).all()

# 5. Display the original weighting method
print("=== Class-balanced Focal Loss ===")

display(
    pd.DataFrame({
        "Emotion": labels,
        "Samples": class_counts,
        "Class Weight": class_weights
    }).round(4)
)

print("\nFocal Loss gamma: 1.5")
print("FocalTrainer defined successfully.")
print("No model training performed.")
```

```python
# STEP 20 | Six-model Test Inference & Candidate Export
# Original competition experiment executed in Google Colab

import numpy as np
import pandas as pd
import torch
from pathlib import Path
from transformers import AutoModelForSequenceClassification

# 1. Set the original Colab experiment directory
root = Path("/content/drive/MyDrive/DM2026_Lab2_NewExperiments")

assert torch.cuda.is_available(), "This step requires the Colab GPU."

# 2. Load previously saved model probabilities
with np.load(
    root / "old_three_model_test_probs.npz",
    allow_pickle=False
) as data:
    old_ids = data["post_id"].astype(str)
    old_labels = data["labels"].astype(str)
    old_probs = data["old_ensemble"].astype(np.float64)

with np.load(
    root / "new_two_model_test_probs.npz",
    allow_pickle=False
) as data:
    new_ids = data["post_id"].astype(str)
    new_labels = data["labels"].astype(str)
    new_probs = data["new_ensemble"].astype(np.float64)

assert np.array_equal(old_ids, new_ids)
assert np.array_equal(old_labels, new_labels)
assert np.array_equal(old_labels, np.asarray(labels))
assert old_probs.shape == new_probs.shape == (1440, 6)

assert np.array_equal(
    old_ids,
    test_clean["post_id"].astype(str).to_numpy()
)

# 3. Load the trained Focal Loss model
if "focal_trainer" in globals():
    focal_model = focal_trainer.model
else:
    checkpoint_path = Path(
        "/content/goemotions_focal_training/checkpoint-168"
    )

    assert checkpoint_path.exists()

    focal_model = (
        AutoModelForSequenceClassification.from_pretrained(
            str(checkpoint_path),
            dtype=torch.float32
        )
    )

focal_model = focal_model.to("cuda")
focal_model.eval()

assert [
    focal_model.config.id2label[i]
    for i in range(6)
] == labels

# 4. Predict test probabilities using the Focal Loss model
texts = test_clean["text"].astype(str).tolist()
focal_batches = []

for start in range(0, len(texts), 16):
    batch = texts[start:start + 16]

    inputs = emotion_tokenizer(
        batch,
        padding=True,
        truncation=True,
        max_length=128,
        return_tensors="pt"
    ).to("cuda")

    with torch.inference_mode():
        logits = focal_model(**inputs).logits
        probabilities = torch.softmax(
            logits.float(),
            dim=-1
        )

    focal_batches.append(probabilities.cpu().numpy())

focal_probs = np.concatenate(
    focal_batches,
    axis=0
).astype(np.float64)

assert focal_probs.shape == (1440, 6)
assert np.isfinite(focal_probs).all()
assert np.allclose(focal_probs.sum(axis=1), 1, atol=1e-4)

# 5. Apply the original ensemble weights
five_probs = (
    0.40 * old_probs
    + 0.60 * new_probs
)

six_probs = (
    0.85 * five_probs
    + 0.15 * focal_probs
)

assert six_probs.shape == (1440, 6)
assert np.allclose(six_probs.sum(axis=1), 1, atol=1e-4)

six_predictions = np.asarray(labels)[
    six_probs.argmax(axis=1)
]

# 6. Align predictions with Kaggle submission IDs
sample_col = "id" if "id" in sample.columns else "post_id"
sample_ids = sample[sample_col].astype(str).to_numpy()

assert len(sample_ids) == 1440
assert set(sample_ids) == set(old_ids)

candidate = (
    pd.DataFrame({
        "id": old_ids,
        "label": six_predictions
    })
    .set_index("id")
    .loc[sample_ids]
    .reset_index()
)

assert candidate.shape == (1440, 2)
assert candidate["id"].is_unique
assert candidate["label"].isin(labels).all()

# 7. Save the original six-model candidate
candidate.to_csv(
    root / "submission_six_focal15.csv",
    index=False
)

np.savez_compressed(
    root / "six_focal_test_probs.npz",
    post_id=old_ids,
    labels=np.asarray(labels),
    focal=focal_probs.astype(np.float32),
    six_ensemble=six_probs.astype(np.float32)
)

# 8. Display results
print("=== Six-model Test Predictions ===")
print("Samples:", len(candidate))

print("\nPrediction Distribution:")
display(candidate["label"].value_counts().reindex(labels))

print("\nSix-model candidate generated successfully.")
print("Exact-match correction is applied in the next step.")
```
